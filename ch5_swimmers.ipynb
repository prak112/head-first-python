```python
import os
import webbrowser
import swimclub

swimdata_files = os.listdir(swimclub.DataFile.DATA_DIR.value)
swimdata_files.remove('.DS_Store')
```

```python
# Using Lists for Sequential Search
swimmer_files = []
for file in swimdata_files:
  name, *_ = swimclub.process_swim_data(file)
  if name not in swimmer_files:
    swimmer_files.append((name, file))

print(len(swimmer_files), "Swimmer Files")
swimmer_files[:5]
```

```python
for n, i in enumerate(swimmer_files):
  if i[0] == "Ruth":
    html = swimclub.generate_bar_chart(i[1])
    webbrowser.open(f"file://{html}")
    print(n, i)
```

```python
# using Dictionaries for Lookup search
swimmers = {}

for file in swimdata_files:
  name, *_ = swimclub.process_swim_data(file)
  if name not in swimmers:
    swimmers[name] = []
  swimmers[name].append(file)  

print(swimmers["Abi"])

for file in swimmers["Ruth"]:
  html = swimclub.generate_bar_chart(file)
  webbrowser.open(f"file://{html}")
  
```

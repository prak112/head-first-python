```python
# Optional: constants declaration using enum class syntax
from enum import Enum
class DataFile(Enum):
  DIRNAME = "swimdata/"
  FNAME = "Darius-13-100m-Fly.txt"

print(f"DataFile Enum Members or CONSTANTS:\n{list(DataFile)}")
```

```python
# sub-tasks 2a, 2b, 2c: Read file and split data for a list of items
with open(DataFile.DIRNAME.value + DataFile.FNAME.value, mode='r') as file:
  '''
  Alternative for file.readlines() is file.readline().
  file.readlines() -> returns list[str until EOF or newline], which is redundant since we need processed data in a list
  file.readline() -> returns str until EOF, which is easier to process and collect as data in a list
  '''
  times = file.readline().strip('\n').split(',')
  print(times)

```

```python
# sub-task 2f: unpacking data from filename using string object methods into variables
swimmer, age_group, distance, stroke = DataFile.FNAME.value.removesuffix(".txt").split("-")
print(f"Swimmer: {swimmer}\nAge Group: {age_group}\nDistance: {distance}\nStroke: {stroke}")
```

```python

# sub-tasks 2d: Convert each of the times in times to number, preferably milliseconds, from string in 'min:sec:msec' format
# sub-task 2e: Calculate average time from the times and represent average in 'min:sec:msec' format
# Tip: always best to convert numerical times to smallest unit for ease of calculations
import statistics

times_in_msec = []
print("\nConverted values from timer format -> milliseconds:")

for t in times:
  minutes, rest = t.split(":")
  seconds, millisecs = rest.split(".")
  times_in_msec.append((int(minutes) * 60 * 100) + (int(seconds) * 100) + int(millisecs))
  print(t, "->", times_in_msec[times.index(t)])
avg_time = round(statistics.mean(times_in_msec) / 100, 2)
mins_secs, millisecs = str(avg_time).split(".")
mins = int(mins_secs) // 60
secs = int(mins_secs) - (mins * 60) # subtract acquired minutes (in seconds) from total seconds for actual representation of seconds, i.e., between 0 to 60
avg_time_in_timerformat = f"{mins}:{secs}.{millisecs}"

print(f"\nAverage time: {avg_time_in_timerformat} -> {round(statistics.mean(times_in_msec))}")
```

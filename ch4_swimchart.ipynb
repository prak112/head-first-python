- Total laps per swimmer not consistent, i.e., lap time must be iterated using `body` variable
- Use `utils.normalize` to adjust lap time values (in millisecs) to be within svg width range (400)

```python
import os
import swimstats
import hfp_utils

CHARTS_PATH = "charts/"
if not os.path.exists(CHARTS_PATH):
  os.makedirs(CHARTS_PATH)

swimdata_files = os.listdir(swimstats.DataFile.DIRNAME.value)
swimdata_files.remove('.DS_Store')
for fname in swimdata_files:
  swimmer, age_group, distance, stroke, times, times_in_msec, avg_in_timerformat = swimstats.process_swim_data(fname)

  html = ""
  title = f"{swimmer}(Under {age_group}) {distance} - {stroke}"

  header = f"""
  <!DOCTYPE html>
  <html>
      <head>
          <title>
              {title}
          </title>
      </head>
      <body>
          <h3>{title}</h3>
  """
  html += header

  time_min = 0
  time_max = max(times_in_msec)
  svg_min = 0
  svg_max = 400
  for t in times_in_msec:
    bar_width = hfp_utils.normalize(t, time_min, time_max, svg_min, svg_max)
    body = f"""
          <svg height="30" width="{svg_max}">
            <rect height="30" width="{bar_width}" style="fill:rgb(0,0,255);" />
          </svg>{times[times_in_msec.index(t)]}<br />
    """
    html += body

  footer = f"""
          <p>Average time: {avg_in_timerformat}</p>
      </body>
  </html>
  """
  html += footer

  # write generated html string to file
  with open(f"{CHARTS_PATH}{title}.html", mode="w") as hf:
    print(html, file=hf) # file argument sends print data hf file
```

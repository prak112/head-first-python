```python
'''
TASK 1 - Unpack the data from the filename into variables for annotating swimming duration of each swimmer by their data.
'''
fname = "Darius-13-100m-Fly.txt"
```

```python
# quick check on string object attributes 
dunder_methods = [] # method names with double underscore a.k.a "magic methods"
general_methods = []

for methd_name in dir(fname):
  if methd_name.__contains__('__'):
    dunder_methods.append(methd_name)
  else:
    general_methods.append(methd_name)
print("Dunder Methods:\n", dunder_methods)
print("General Methods:\n", general_methods)
```

```python
"""
Problem - '.txt' is not required in the categories list.
Actual output:  ['Darius', '13', '100m', 'Fly.txt']
Expected output: ['Darius', '13', '100m', 'Fly']
"""
# Alternate Solution
"""
'strip' method breaks the input string based on the seperator and returns a string object
'split' method breaks the input string into individual string objects based on the seperator, collects it in a list and returns it.
"""
type(fname.strip(".txt")) # strip output: 'Darius-13-100m-Fly'
```

```python
# Book Solution
""" 
Explanation - 'removesuffix' is technically and logically appropriate to the given problem than 'strip' which is more applicable to removing whitespaces or specific characters in an input string.
"""
print("Method output: ", fname.removesuffix(".txt"))
print("Manual output: ", fname[:-len(".txt")]) # logic of 'removesuffix' method
```

```python
fname.removesuffix(".txt").split("-")
```

```python
# unpacking the data from filename using string object methods into variables
swimmer, age_group, distance, stroke = fname.removesuffix(".txt").split("-")

print(f"Swimmer: {swimmer}\nAge Group: {age_group}\nDistance: {distance}\nStroke: {stroke}")
```
